```python
def power_set(a):
    if a == []:
        return [[]]
    P= [[]]
    for p in power_set(a[1:]):
        P.append([a[0]] + p)
    for p in power_set(a[1:]):
        if p != []:
            P.append(p)
    return P
```

```python
# CODE ĐÃ RÚT GỌN MÀ KHÔNG HAY ĐỔI THUẬT TOÁN
def power_set_2(a):
    if not a:
        return [[]]
    rest = power_set_2(a[1:])
    return [[]] + [[a[0]] + p for p in rest] + [p for p in rest if p]
```
