```python
def combinations(a,r):
    if r==1: 
        return [[i] for i in a]
    n = len(a)
    if r == n:
        return [a]
    C=[]
    for c in combinations(a[1:], r-1):
        c= [a[0]] + c
        C.append(c)
    for c in combinations(a[1:], r):
        C.append(c)
    return C

combinations([1,2,3,4,5], 3)
```
