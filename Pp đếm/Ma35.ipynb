```python
def next_combinations(n , a):
    r = len(a)
    i = r-1
    while i>= 0 and a[i] == n- r +(i+1):
        i -= 1
    if i == -1:
        return None
    return a[:i] + [a[i] +j for j in range(1, r-i +1)]
```

```python
def next_combinations(n , a):
    r = len(a)             # r là số phần tử trong tổ hợp hiện tại
    i = r - 1              # i bắt đầu từ phần tử cuối cùng (duyệt ngược)

    # Bước 1: Tìm vị trí i mà tại đó a[i] có thể tăng thêm được
    # Điều kiện a[i] == n - r + (i + 1) là khi a[i] đã đạt giá trị tối đa có thể tại vị trí đó
    while i >= 0 and a[i] == n - r + (i + 1):
        i -= 1

    # Nếu không còn vị trí nào có thể tăng ⇒ đây là tổ hợp cuối cùng
    if i == -1:
        return None

    # Bước 2: Tăng phần tử a[i] lên 1 và cập nhật các phần tử sau nó theo quy tắc tổ hợp
    # Ví dụ: nếu a = [2, 3, 5] → i = 1, a[1] = 3 → a[1] = 4, các phần sau là [5, 6]
    return a[:i] + [a[i] + j for j in range(1, r - i + 1)]
```
