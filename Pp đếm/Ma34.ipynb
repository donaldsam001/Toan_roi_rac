```python
def next_permutations(a):
    n = len(a)
    k = n-1
    while k >= 1 and a[k-1] > a[k] :
        k -= 1
    if k == 0:
        return None
    i = n-1
    while a[i] < a[k-1]:
        i -= 1
    a[k-1], a[i] = a[i], a[k-1]

    b= a[k:]
    b.reverse()
    return a[:k] + b
```

```python
def next_permutations(a):
    n = len(a)
    k = n - 1

    # Bước 1: Tìm vị trí k sao cho a[k-1] < a[k]
    while k >= 1 and a[k - 1] > a[k]:
        k -= 1

    # Nếu không tìm được (tức là dãy đang giảm dần toàn bộ), không có hoán vị kế tiếp
    if k == 0:
        return None

    # Bước 2: Tìm phần tử a[i] lớn hơn a[k-1], từ phải qua trái
    i = n - 1
    while a[i] < a[k - 1]:
        i -= 1

    # Bước 3: Hoán đổi a[k-1] và a[i]
    a[k - 1], a[i] = a[i], a[k - 1]

    # Bước 4: Đảo ngược phần phía sau vị trí k để đảm bảo nhỏ nhất
    b = a[k:]       # Lấy phần sau vị trí k
    b.reverse()     # Đảo ngược phần này để có hoán vị kế tiếp nhỏ nhất
    return a[:k] + b  # Ghép phần đầu với phần đã đảo để tạo hoán vị kế tiếp
```
