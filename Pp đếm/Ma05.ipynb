```python
def matrix_mul(A,B):
    m, n = len(A), len(B)
    p = len(B[0])
    C = [[ 0 for j in range(p)] for i in range(m)]
    for i in range(m):
        for j in range(p):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]

    return C
```

```python
matrix_mul(A, B)
```
