```python
def combinations_with_replacement( a, r):
    n = len(a)
    if n == 1:
        return [a*r]
    if r == 1:
        return [[i] for i in a]
    C= []
    for c in combinations_with_replacement(a, r-1):
        c= [a[0]]+ c
        C.append(c)
    for c in combinations_with_replacement(a[1:], r):
        C.append(c)
    return C
    
```

```python
combinations_with_replacement(['A', 'B', 'C'], 4)
```
