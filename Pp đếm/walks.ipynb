```python
# cách đi từ X(a,b) -> Y(x,y)
def walks(a, b, x, y):
    if a == x:
        return ['U' * (y - b)]
    if b == y:
        return ['R' * (x - a)]
    W = []
    for w in walks(a + 1, b, x, y):
        w = 'R' + w
        W.append(w)
    for w in walks(a, b + 1, x, y):
        w = 'U' + w
        W.append(w)
    return W
```
