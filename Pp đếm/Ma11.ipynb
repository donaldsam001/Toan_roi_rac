```python
def permutations(a):
    n = len(a)
    if n == 1:
        return [a]
    P = []
    for i in range(n):
        b = a.copy()
        x = b.pop(i)
        for p in permutations(b):
            p = [x] + p
            P.append(p)
    return P
```

```python
permutations([1, 2, 3])
```

```python
def permutations(a):
    n = len(a)                    # Độ dài danh sách a
    if n == 1:                    # Trường hợp cơ sở: nếu danh sách chỉ còn 1 phần tử
        return [a]                # Trả về danh sách chứa 1 hoán vị (chính nó)

    P = []                        # Danh sách để lưu tất cả các hoán vị
    for i in range(n):            # Lặp qua từng vị trí trong danh sách
        b = a.copy()              # Tạo bản sao để không thay đổi danh sách gốc
        x = b.pop(i)              # Lấy phần tử tại vị trí i ra (x là phần tử cố định đầu tiên của hoán vị)
        for p in permutations(b): # Gọi đệ quy để tạo hoán vị cho phần còn lại
            p = [x] + p           # Ghép phần tử x vào đầu mỗi hoán vị còn lại
            P.append(p)           # Thêm vào danh sách kết quả
    return P                      # Trả về toàn bộ danh sách hoán vị
```

```python
permutations([1, 2, 3])
```
