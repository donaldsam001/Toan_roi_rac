```python
def binary_strs(n):
    if n == 1:
        return ['0', '1']
    A = []
    for s in binary_strs(n - 1):
        A.append('0' + s)
    for s in binary_strs(n - 1):
        A.append('1' + s)
    return A

binary_strs(3)
```
