```python
from sympy import *
binomial(10, 4)
```

```python
def binomial(n ,r):
    p= 1
    for i in range(r):
        p = p*(n-i)// (i+1)
    return p
```

```python
binomial(10, 4)
```

```python
def binomial(n,r):
    if r == 0 or r ==n:
        return 1
    return binomial(n-1, r-1)+ binomial(n-1, r)
binomial(10,4)
```

```python
def binomial(n, r): # To hop chập 4 của 10
    a= [1]
    for i in range( 1, n +1):
        for j in range(i -1, 0,-1):
            a[j] += a[j-1]
        a.append(1)
    return a[r]
binomial(10, 4)
```
