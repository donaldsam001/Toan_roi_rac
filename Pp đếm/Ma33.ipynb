```python
def compare(a, b):
    m, n = len(a), len(b)
    i = 0 
    while i< m and i< n and a[i] == b[i]:
        i += 1
        if i == m == n:
            return ('=')
        if i == m < n:
            return ('<')
        if i == n < m:
            return ('>')
        if i < m and  i <n:
            if a[i] < b[i]:
                return ('<')
            else:
                return ('>')
```

```python
compare([4,1,2], [4,1,2,3])
```

```python
compare([3,1,4], [3,1,2,5])
```
