```python
def compare(a, b):
    m, n = len(a), len(b)   # m và n là độ dài của hai danh sách a và b
    i = 0                   # Biến i dùng để duyệt qua các phần tử của a và b

    # So sánh từng phần tử a[i] và b[i] từ trái qua phải (tức là theo thứ tự)
    while i < m and i < n and a[i] == b[i]:
        i += 1              # Tăng i nếu các phần tử tại vị trí i bằng nhau

    # Sau vòng lặp:
    # Nếu đã so hết cả hai danh sách và không thấy khác biệt ⇒ chúng bằng nhau
    if i == m == n:
        return '='          # Trả về '=' nếu a và b giống hệt nhau

    # Nếu đã so hết a (i == m) nhưng b còn phần tử ⇒ a ngắn hơn b
    if i == m < n:
        return '<'          # Trả về '<' nếu a nhỏ hơn b về độ dài (và các phần đầu bằng nhau)

    # Nếu đã so hết b (i == n) nhưng a còn phần tử ⇒ a dài hơn b
    if i == n < m:
        return '>'          # Trả về '>' nếu a lớn hơn b về độ dài (và các phần đầu bằng nhau)

    # Nếu khác biệt xảy ra tại vị trí i, so sánh giá trị a[i] và b[i]
    if i < m and i < n:
        if a[i] < b[i]:
            return '<'      # Nếu phần tử a nhỏ hơn b tại vị trí i
        else:
            return '>'      # Nếu phần tử a lớn hơn b tại vị trí i
```
