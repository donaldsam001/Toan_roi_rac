```python
from sympy import *

# Define symbols
n, P, r, S = symbols('n P r S')
a = symbols('a', cls=Function)

# Solve the recurrence relation
sol = rsolve(
    -a(n) + (1 + r) * a(n - 1) - P,
    a(n),
    {a(0): S}
)

# Simplify the solution
sol = sol.simplify()

# Solve for P
sol_P = solve(sol, P)

# Substitute values into the solution
result = sol_P[0].subs({S: 100, r: 0.01, n: 12})

# Display the result
result
```

```python
def summands(n):
    if n == 1:
        return [[1]]
    S = []
    for s in summands(n - 1):
        s[0] += 1
        S.append(s)
    for s in summands(n - 1):
        s = [1] + s
        S.append(s)
    return S

summands(3)
```

```python
def BubbleSort(x):
    n = len(x)
    if n == 1:
        return x
    for i in range(n - 1, 0, -1):
        if x[i] < x[i - 1]:
            x[i], x[i - 1] = x[i - 1], x[i]
    return [x[0]] + BubbleSort(x[1:])

BubbleSort([7, 9, 2, 5, 8])
```

```python
def hanoi_tower(n, A, B, C):
    if n == 1:
        return [[1, A, B]]
    return hanoi_tower(n - 1, A, C, B) + [[n, A, B]] + hanoi_tower(n - 1, C, B, A)

hanoi_tower(3, 'A', 'B', 'C')
```

```python
def quaternary_strs(n):
    if n == 1:
        return [[i] for i in range(4)]
    S = []
    for i in range(4):
        for s in quaternary_strs(n - 1):
            s = [i] + s
            S.append(s)
    return S

quaternary_strs(2)
```

```python
def quaternary_strs_1s_even(n):
    if n == 1:
        return [[0], [2], [3]]
    S = []
    for s in quaternary_strs(n - 1):
        if s not in quaternary_strs_1s_even(n - 1):
            s = [1] + s
            S.append(s)
    for i in [0, 2, 3]:
        for s in quaternary_strs_1s_even(n - 1):
            s = [i] + s
            S.append(s)
    return S

quaternary_strs_1s_even(2)
```

```python
from sympy import *

n = symbols('n')
a = symbols('a', cls=Function)

# Solve the recurrence relation without initial conditions
solution1 = rsolve(
    -a(n) + 3 * a(n - 2) - 2 * a(n - 3),
    a(n)
)
print("Solution without initial conditions:", solution1)

# Solve the recurrence relation with initial conditions
solution2 = rsolve(
    -a(n) + 3 * a(n - 2) - 2 * a(n - 3),
    a(n),
    {a(0): 5, a(1): -1, a(2): 2}
)
print("Solution with initial conditions:", solution2)
```

```python
from sympy import *

# Define symbols
x = symbols('x')
C1, C2, C3 = symbols('C1 C2 C3')
n = symbols('n')

# Define polynomial and factorize
P = x**3 - 3*x + 2
P_factored = P.factor()

# Define the expression
ans = C1 + C2*n + (-2)**n * C3

# Substitute values into the expression
substituted_values = [ans.subs(n, i) for i in [0, 1, 2]]

# Create equations and solve
eqns = [ans.subs(n, i) - a for i, a in zip([0, 1, 2], [5, -1, 2])]
solution = solve(eqns)

# Display results
P_factored, substituted_values, solution
```

```python
from sympy import *

# Define symbols
n = symbols('n', integer=True)
a = symbols('a', cls=Function)

# Solve the recurrence relation
ans = rsolve(
    a(n + 1) - 2 * a(n) + 2 * a(n - 1),
    a(n),
    {a(0): 1, a(1): 2}
)

# Define polar function
polar = lambda z: abs(z) * E**(I * arg(z))

# Display the solution
print(ans)

# Update the solution with polar terms
ans = (Rational(1, 2) + I / 2) * polar(1 - I)**n + (Rational(1, 2) - I / 2) * polar(1 + I)**n

# Display the updated solution
ans

# Extract the real part of the solution
re(ans)
```

```python
from sympy import *

# Define symbols
x = symbols('x')
n = symbols('n', integer=True)
C1, C2 = symbols('C1 C2')

# Define polynomial and solve
P = x**2 - 2*x + 2
roots = solve(P)

# Define complex number and its polar form
x = 1 + I
r = abs(x)
phi = arg(x)

# Define the expression
ans = r**n * (C1 * cos(n * phi) + C2 * sin(n * phi))

# Substitute values into the expression
substituted_values = [ans.subs(n, i).simplify() for i in [0, 1]]

# Create equations and solve
eqns = [ans.subs(n, i).simplify() - a for i, a in zip([0, 1], [1, 2])]
solution = solve(eqns)

# Display results
roots, substituted_values, solution
```

```python
def subsets_with_condition(n):
    if n == 1:
        return [[], [1]]
    if n == 2:
        return [[], [1], [2]]
    S = []
    for s in subsets_with_condition(n - 1):
        S.append(s)
    for s in subsets_with_condition(n - 2):
        s = s + [n]
        S.append(s)
    return S

subsets_with_condition(3)
```

```python
def binary_strs(n):
    if n == 1:
        return ['0', '1']
    if n == 2:
        return ['01', '10', '11']
    S = []
    for s in binary_strs(n - 2):
        s = '01' + s
        S.append(s)
    for s in binary_strs(n - 1):
        s = '1' + s
        S.append(s)
    return S

binary_strs(3)
```

```python
from sympy import *

n = symbols('n')
a = symbols('a', cls=Function)

# Solve the recurrence relation
ans = rsolve(
    a(n) - 2 * a(n - 2),
    a(n),
    {a(1): 1, a(2): 2}
)
print(ans)

# Define a new symbol k
k = symbols('k', integer=True)

# Substitute even and odd values of n
even_substitution = ans.subs(n, 2 * k).simplify()
odd_substitution = ans.subs(n, 2 * k + 1).simplify()

print(even_substitution)
print(odd_substitution)
```

```python
def symmetric_summands(n):
    if n == 1:
        return [[1]]
    if n == 2:
        return [[2], [1, 1]]
    S = []
    for s in symmetric_summands(n - 2):
        s[0] += 1
        s[-1] += 1
        S.append(s)
    for s in symmetric_summands(n - 2):
        s = [1] + s + [1]
        S.append(s)
    return S

symmetric_summands(4)
```
