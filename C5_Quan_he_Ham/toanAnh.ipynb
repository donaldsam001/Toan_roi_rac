```python
def S(a: list, n: int):
    m = len(a)
    if n == 1:
        return [[a]]
    if n == m:
        return [[[x] for x in a]]
    S_out = []
    x = a[-1]  # x là a[n-1]
    for part in S(a[:-1], n - 1):
        part.append([x])
        S_out.append(part)
    for part in S(a[:-1], n):
        for i in range(len(part)):
            part_copy = []  # dùng part_copy = part.copy()
            for item in part:
                part_copy.append(item.copy())
            part_copy[i].append(x)
            S_out.append(part_copy)
    return S_out

S([1, 2, 3], 2)
```
