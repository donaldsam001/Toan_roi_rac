```python
'''
Cho số nguyên dương n. Chứng minh số tổng riêng của n là 2n−1. [Gợi
ý: xét số hạng đầu của mỗi tổng riêng là 1 hoặc khác 1.]
'''
```

```python
def summands(n):
    if n == 1:
        return [[1]]
    L = []
    for x in summands(n - 1):
        y = x.copy()
        x.append(1)
        L.append(x)
        y[-1] += 1
        L.append(y)
    return L
```
