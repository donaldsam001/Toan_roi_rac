```python
def factorint(n: int) -> dict:
    d = {}
    i = 2
    while n > 1:
        while n % i == 0:
            if i in d:
                d[i] += 1
            else:
                d[i] = 1
            n //= i
        i += 1
    return d
```

```python
factorint(100)
```

```python
def display_factors(n: int, b: int) -> list:
    a= []
    while n!= 0:
        a.append(n % b)
        n //= b
    a.reverse()
    return a
```

```python
display_factors(12345, 8)
```

```python
def solve(a: list, b: int) -> int:
    k = len(a)
    n = a[k - 1]
    for i in range(k - 2, -1, -1):
        n = b * n + a[i]
    return n
```

```python
solve([1,7,0,0,3], 8)
```

```python
a = [2, 4, 3]
b = [3, 2, 1, 4]
base = 5

k = len(a)
l = len(b)
s = [0] * (k + l)

for i in range(l):
    r = R = 0
    for j in range(k):
        t = a[j] * b[i] + r
        p = t % base
        r = t // base
        t = s[i + j] + p + R
        s[i + j] = t % base
        R = t // base
    s[k + i] = r + R

s
```
