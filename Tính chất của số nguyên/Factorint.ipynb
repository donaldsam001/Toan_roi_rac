```python
from sympy import *
factorint (980220) # {2: 2, 3: 1, 5: 1, 17: 1, 31: 2}
```

```python
def factorint(n: int) -> dict:
    d = {}
    i = 2
    while n > 1:
        while n % i == 0:
            if i in d:
                d[i] += 1
            else:
                d[i] = 1
            n //= i
        i += 1
    return d
```

```python
factorint(980220)
```

```python
def display_factors(n: int, b: int) -> list:
    a= []
    while n!= 0:
        a.append(n % b)
        n //= b
    a.reverse()
    return a
```

```python
display_factors(12345, 8)
```

```python
def solve(a: list, b: int) -> int:
    k = len(a)
    n = a[k - 1]
    for i in range(k - 2, -1, -1):
        n = b * n + a[i]
    return n
```

```python
solve([1,7,0,0,3], 8)
```

```python
# Tính 342(5) × 4213(5)
a = [2, 4, 3]
b = [3, 2, 1, 4]
base = 5

k = len(a)
l = len(b)
s = [0] * (k + l)

for i in range(l):
    r = R = 0
    for j in range(k):
        t = a[j] * b[i] + r
        p = t % base
        r = t // base
        t = s[i + j] + p + R
        s[i + j] = t % base
        R = t // base
    s[k + i] = r + R

s
```

```python
# Tính 7246(8) + 4735(8) (cơ số 8)
a = [6, 4, 2, 7]
b = [5, 3, 7, 4]
base = 8
k = len(a)  # = len(b), hơn k lý thuyết 1 đơn vị
r = 0
s = [0] * (k + 1)

for i in range(k):
    t = a[i] + b[i] + r
    s[i] = t % base
    r = t // base

s[k] = r
s  # → [3, 0, 2, 4, 1]
```
