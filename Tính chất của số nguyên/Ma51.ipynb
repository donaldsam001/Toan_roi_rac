```python
def gcdex(a,b): #  khai triển Euclid mở rộng của 91 và 287
    x0, y0 = 1, 0
    x1, y1 = 0, 1
    while b!= 0:
        q= a//b
        a,b = b, a%b
        x= x0 - x1 *q
        y= y0 -y1 *q
        x0, y0 = x1, y1
        x1, y1 =x, y
    return x0, y0, a
```

```python
gcdex(91, 287)
```
