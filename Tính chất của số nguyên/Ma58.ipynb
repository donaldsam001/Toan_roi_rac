```python
a = [6, 4, 2, 7]  # Đại diện cho số A trong hệ cơ số 8
b = [5, 3, 7, 4]  # Đại diện cho số B trong hệ cơ số 8
base = 8

k = len(a)
r = 0
s = [0] * (k + 1)  # Danh sách kết quả (có thể có thêm 1 chữ số nếu có nhớ cuối)

for i in range(k):
    t = a[i] + b[i] + r   # Cộng hai chữ số cùng vị trí và số nhớ
    s[i] = t % base       # Lưu chữ số hiện tại sau khi lấy phần dư
    r = t // base         # Cập nhật số nhớ (carry)

s[k] = r  # Nếu còn nhớ sau cùng, lưu vào phần tử cuối
s
```
