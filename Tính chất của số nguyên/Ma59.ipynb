```python
a = [2, 4, 3]       # Số thứ nhất trong hệ cơ số 5, biểu diễn từ chữ số thấp đến cao (3 là chữ số hàng cao nhất)
b = [3, 2, 1, 4]    # Số thứ hai trong hệ cơ số 5, cũng biểu diễn từ thấp đến cao
base = 5            # Cơ số đang sử dụng

k = len(a)          # Độ dài của số a (3 chữ số)
l = len(b)          # Độ dài của số b (4 chữ số)

s = [0] * (k + l)   # Mảng kết quả, có độ dài tối đa là k + l (số chữ số tối đa của tích hai số)

# Duyệt qua từng chữ số của b (giống như nhân giấy với từng chữ số của b)
for i in range(l):
    r = R = 0       # r: số nhớ của phép nhân, R: số nhớ của phép cộng
    for j in range(k):
        t = a[j] * b[i] + r     # Nhân chữ số a[j] với b[i], cộng thêm số nhớ r từ bước trước
        p = t % base            # p là chữ số kết quả sau khi lấy phần dư theo cơ số
        r = t // base           # Cập nhật số nhớ của phép nhân

        t = s[i + j] + p + R    # Cộng chữ số p vào vị trí s[i + j], cộng thêm số nhớ R
        s[i + j] = t % base     # Cập nhật chữ số tại vị trí i + j
        R = t // base           # Cập nhật số nhớ mới của phép cộng

    # Sau khi hoàn thành vòng lặp j (tức là đã nhân xong a với b[i]), cộng số nhớ cuối cùng vào s[k+i]
    s[k + i] = r + R            # Có thể thêm: s[k+i] += r + R nếu cộng dồn kết quả từ nhiều vòng lặp
s
```
