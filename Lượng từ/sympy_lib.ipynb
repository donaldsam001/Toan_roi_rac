```python

x = symbols('x')
expr = x**2 + 3*x + 2
expr_substituted = expr.subs(x, 2)

print(expr_substituted)  # Output: 12
```

```python

x, y = symbols('x y')
expr = x**2 + y
expr_sub = expr.subs({x: 3, y: 5})
print(expr_sub)  # Output: 14
```

```python
n, i, j, k = symbols('n i j k')

expr = summation(1, (k, j, n), (j, i, n), (i, 1, n)) - summation(1, (k, i, n), (i, 1, n)) - summation(1, (j, i, n), (i, 1, n))
expr.simplify()
```
